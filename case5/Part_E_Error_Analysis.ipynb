```python
import pandas as pd

# Load BERT errors
fp = pd.read_csv("BERT_False_Positives.csv")
fn = pd.read_csv("BERT_False_Negatives.csv")

# Select 5 examples from each
fp_selected = fp.iloc[[2, 4, 6, 9, 10]].copy()
fn_selected = fn.iloc[[0, 1, 2, 3, 6]].copy()

# Add error types
fp_selected["error_type"] = "False Positive"
fn_selected["error_type"] = "False Negative"

# Combine into one table
error_table = pd.concat(
    [fp_selected, fn_selected],
    ignore_index=True
)

error_table
```

```python
explanations = [
    "The job advertisement includes salary and contact information, making it resemble promotional spam.",
    "The message resembles a mobile plan advertisement because it mentions texting and calling limits.",
    "The unusual formatting and request for a response may resemble mass-sent promotional messages.",
    "The motivational message includes a third-party SMS service name, which may make it appear automated.",
    "The mention of a free messaging service may resemble language used in promotional messages.",
    "The message is extremely short and lacks enough context to clearly identify it as spam.",
    "The message resembles an ordinary joke or news headline rather than a typical spam advertisement.",
    "The message sounds like someone describing their experience rather than directly advertising a service.",
    "The message starts like a personal conversation, potentially obscuring the promotional link and paid messaging details.",
    "The message resembles a routine content notification rather than an explicit spam offer."
]

error_table["possible_explanation"] = explanations

# Organize columns
error_table = error_table[
    ["error_type", "text", "label", "predicted_label", "possible_explanation"]
]

pd.set_option("display.max_colwidth", None)
error_table
```

We analyzed five false positives and five false negatives from BERT, which had the highest spam F1 score (0.9338). The false positives often contained promotional language, unusual formatting, or references to messaging services, which may have made legitimate messages resemble spam. The false negatives included short messages with little context and spam messages written like normal conversations or notifications. These results suggest that BERT can still struggle with ambiguous wording and messages that do not follow typical spam patterns.
