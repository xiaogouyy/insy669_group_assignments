## Import libraries and load data set

```python
import pandas as pd
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

df = pd.read_excel("../data/Case3_Data1.xlsx")
df.head()
```

## Apply VADER

```python

analyzer = SentimentIntensityAnalyzer()

# Light preprocessing: only remove extra surrounding whitespace
df["vader_text"] = df["review_text"].fillna("").str.strip()

# Generate compound sentiment scores
df["vader_compound"] = df["vader_text"].apply(
    lambda text: analyzer.polarity_scores(text)["compound"]
)

df[["reviewer_id", "review_text", "star_rating", "vader_compound"]].head()
```

## Stats

```python

# Average VADER sentiment for each star rating
df.groupby("star_rating")["vader_compound"].agg(
    ["count", "mean"]
).round(3)
```

## Save data

```python

df[["reviewer_id", "vader_compound"]].to_csv(
    "../results/person2_vader_scores.csv",
    index=False
)
```

## Report

For this approach, we used VADER to analyze the sentiment of each review. We kept preprocessing to a minimum and only removed extra whitespace while preserving capitalization, punctuation, and negations, since VADER uses these to assign sentiment values. Each review was then assigned a compound score between -1 (negative) to +1 (positive). We then compared these scores with the actual star ratings to see how well VADER's sentiment scores reflected the ratings customers gave.

The VADER results show a clear relationship between sentiment scores and star ratings. Reviews with 1 star had the lowest average sentiment score (-0.283), while 5-star reviews had the highest (0.683). The average sentiment score increased with each star rating, suggesting that VADER was able to capture how positively or negatively customers felt in their reviews. However, one thing we noticed was that the average score for 2-star reviews was slightly positive (0.057), showing that VADER doesn't always fully reflect the rating a customer gives.
